```python
import pandas as pd
import requests
import matplotlib.pyplot as plt
import requests
```

```python
# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/children-born-per-woman.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/children-born-per-woman.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()
```

```python
mexico = df[df['entity'] == 'Mexico']   
print(mexico)
```

```python
df.head(10)
```

```python
japon = df[df['entity'] == 'Japan']
print(japon)
```

```python
anios_mexico = mexico['year'].to_numpy()
tasa_mexico = mexico['fertility_rate_hist'].to_numpy()

anios_japon = japon['year'].to_numpy()
tasa_japon = japon['fertility_rate_hist'].to_numpy()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(
    anios_mexico, 
    tasa_mexico, 
    label='Mexico', 
    color='blue'
)

plt.title('Fertility Rate in Mexico and Japan (1960-2020)')
plt.xlabel('Year')
plt.ylabel('Fertility Rate (children per woman)')
plt.grid(True)
plt.legend()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(
    anios_japon, 
    tasa_japon,
    label='Japan', 
    color='blue'
)

plt.title('Fertility Rate in Mexico and Japan (1960-2020)')
plt.xlabel('Year')
plt.ylabel('Fertility Rate (children per woman)')
plt.grid(True)
plt.legend()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(
    anios_japon, 
    tasa_japon,
    label='Japan', 
    color='blue'
)

plt.plot(
    anios_mexico, 
    tasa_mexico, 
    label='Mexico', 
    color='red'
)

plt.title('Fertility Rate in Mexico and Japan (1960-2020)')
plt.xlabel('Year')
plt.ylabel('Fertility Rate (children per woman)')
plt.grid(True)
plt.legend()
```
