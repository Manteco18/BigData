```python
import pandas as pd
import requests

# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/life-expectancy.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/life-expectancy.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()
```

```python
df.head(3)
```

```python
df.rename(columns={"entity" : "Pais"}, inplace=True)
df.rename(columns={"code" : "Code"}, inplace=True)
df.rename(columns={"code" : "Code"}, inplace=True)
df.rename(columns={"year" : "Anos"}, inplace=True)
df.rename(columns={"esperanza_de_vida" : "Esperanza_de_vida"}, inplace=True)
```

```python
#df.set_index("Pais", inplace=True)
df.reset_index(inplace=True)
```

```python
df.dropna(subset=["Code"], inplace=True)
df.isnull().sum()
```

```python
resultado = (
    df.groupby("Pais")["life_expectancy_0"]
      .mean()#Calcula el promedio de esperanza de vida para cada país.
      .sort_values(ascending=True)# Ordena los resultados de menor a mayor.
)
print(resultado)
```
