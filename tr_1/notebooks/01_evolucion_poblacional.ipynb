```python
columna = df["Annual change in population"]

print("Media:", columna.mean())
print("Mediana:", columna.median())
print("Desviación estándar:", columna.std())

indice_min = columna.idxmin()
indice_max = columna.idxmax()

print("Mínimo:", df.loc[indice_min, ["Entity", "Annual change in population"]])
print("Máximo:", df.loc[indice_max, ["Entity", "Annual change in population"]])
```

```python
#df_minimo = df[["Entity", "Annual change in population"]]
#df_minimo.min(numeric_only=True)

indice = df["Annual change in population"].idxmin()

print(df.loc[indice, ["Entity", "Annual change in population"]])
```

```python
#df_maximo = df[["Entity", "Annual change in population"]]
#df_maximo.max(numeric_only=True)

indice = df[(df["Entity"] != "World") & (df["Entity"] != "Less developed regions" )]["Annual change in population"].idxmax()
print(df.loc[indice, ["Entity", "Annual change in population"]])
```
