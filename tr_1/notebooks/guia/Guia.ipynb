# 2. Filtrar datos

Una de las cosas más importantes del examen.
Filtrar por país

> mexico = df[df["entity"] == "Mexico"]

Filtrar por año

> datos_2023 = df[df["year"] == 2023]

Filtrar por país Y año

> mexico_2023 = df[
    (df["entity"] == "Mexico") &
    (df["year"] == 2023)
]

# 3. .loc vs .iloc
.loc
Se utiliza para seleccionar usando etiquetas/nombres.

> df.loc[5, "valor"]

También:

> df.loc[df["entity"] == "Mexico"]

.iloc
Se utiliza para seleccionar mediante posición numérica.

>df.iloc[0]
Primera fila.

>df.iloc[0, 2]
Primera fila, tercera columna.
Truco para recordar
loc = location/nombre
iloc = índice/posición
